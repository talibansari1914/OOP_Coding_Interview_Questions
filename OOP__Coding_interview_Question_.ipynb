```python
# 2. Create classes Addition and Concatenation with an operate method to handle
# + operator for numbers and strings.
class Addition:
    def __init__(self,value):
        self.value=value

    def __add__(self,other):
        if isinstance(self.value,(int,float)) and isinstance(other.value,(int,float)):
            return Addition(self.value+other.value)
        else:
            return f"Error:Value should be type of(int,float)"
    def __repr__(self):
        return f"Addition of {self.value}"

class Concatenation:
    def __init__(self,text):
        self.text=text

    def __add__(self,other):
        if isinstance(self.text,str) and isinstance(other.text,str):
            return Concatenation(self.text+other.text)
        else:
            return f"Should be in text or string format"
    def __repr__(self):
        return f"Concatenation ({self.text})"

a1=Addition(20)
a2=Addition(30)
print(a1+a2)
c=Concatenation("Talib")
c_2=Concatenation("Ansari")
```

```python
# 2. Use the abc module to create a DatabaseConnection with abstract connect and
# query methods
from abc import ABC,abstractmethod
class DatabaseConnection(ABC):
    @abstractmethod
    def connect(self):
        pass
    @abstractmethod
    def query(self):
        pass

class MySqlDatabaseConnection(DatabaseConnection):
    def connect(self):
        return f"Connected to the MySql"
    def query(self):
        return f"Query executed in MySql Database"
class SqlLiteDatabaseConnection(DatabaseConnection):
    def connect(self):
        return f"connected to the SqlLite"
    def query(self):
        return f"Query Executed in SqlLiteDatabaseConnection"

def database_connection(objects):
    for obj in objects:
        print(obj.connect())
        print(obj.query())
        print("====="*10)
m=MySqlDatabaseConnection()
s=SqlLiteDatabaseConnection()
objects=[m,s]

database_connection(objects)
```

```python
# 5. Create an abstract Authenticator class with an authenticate method, implement
# in PasswordAuth and BiometricAuth .
from abc import ABC,abstractmethod
class Authenticator(ABC):
    @abstractmethod
    def authenticate(self):
        pass
        
class PasswordAuth(Authenticator):
    def __init__(self,password):
        self.password=password
        
    def authenticate(self):
        if self.password=="password@123":
            return f"Access granted .."
            
class BiometricAuth(Authenticator):
    def __init__(self,press_finger):
        self.press_finger=press_finger
        
    def authenticate(self):
        return f"Press '{self.press_finger}' on the screen to authenticate yourself"

def authentication_password(objects):
    for obj in objects:
        print(obj.authenticate())
        
p=PasswordAuth("password@123")
b=BiometricAuth("Right Thumb")
objects=[p,b]
authentication_password(objects)
        
```

```python
# 7. Create a House class that manages the lifecycle of Room objects.
class Room:
    def __init__(self,room_no,bed_size,light):
        self.room_no=room_no
        self.bed_size=bed_size
        self.light=light
    def __repr__(self):
        return f"room:{self.room_no} ,bed_size:{self.bed_size},Light:{self.light}"
class House:
    def __init__(self,house_name):
        self.house_name=house_name
        self.room_list=[]
    def add_room(self,obj):
        if isinstance(obj,Room):
            self.room_list.append(obj)
        else:
            raise ValuError("Object should be type of Room")
    def remove(self,obj_2):
        if obj_2 in self.room_list:
            self.room_list.remove(obj_2)
        else:
            raise ValueError("Object not found in the list")
    def show(self):
        if  self.room_list:
            return f"For House:{self.house_name} ,{self.room_list}"
        else:
            raise ValueError("Room_list is Empty")
r1=Room("Room-01","13*12","White & Grey")
r2=Room("Room-02","14*15","Blue & Black")
r3=Room("Room-03","12*15","Red & Green")
h=House("Swarg")
h.add_room(r1)
h.add_room(r2)
h.add_room(r3)
h.show()
```

```python
# 8. Create an Order class that composes a Customer and a list of Item objects.
class Order:
    def __init__(self,order_id,order_name,price,quantity):
        self.order_id=order_id
        self.order_name=order_name
        self.price=price
        self.quantity=quantity
    def __repr__(self):
        return f"Order Id:{self.order_id} , Order Item:{self.order_name} , Price:{self.price} rs , Quantity:{self.quantity}"
class Customer:
    def __init__(self,customer_name):
        self.customer_name=customer_name
        self.items=[]
    def add_item(self,obj):
        if obj not in self.items:
            self.items.append(obj)
            print("items added to the order")
        else:
            for o in self.items:
                if o.order_name==obj.order_name:
                    o.quantity+=obj.quantity
                    print(f"Quantity updated for '{o.order_name}' order")
                
    def remove_item(self,obj):
        if obj in self.items:
            self.items.remove(obj)
            print(f"{obj} order cancelled")
        else:
            raise ValueError(f"{obj} not order yet")
    def show(self):
        if  self.items:
            print(f"\n'{self.customer_name}' please confirm your order..")
            print("======="*5)
            for order in self.items:
                print(order)
            total_price=sum(o.price*o.quantity for o in self.items)
            print(f"Total price will be :{total_price} rs")
        else:
            raise ValueError(f"There is no order in the list")
print("===========Accessing the Class info===========")
o1=Order(101,"Pasta",100,2)
o2=Order(102,"Burger",50,1)
o3=Order(103,"Biryani",300,2)
o4=Order(104,"Mutton Curry",500,1)

c=Customer("Sam Curran")
c.add_item(o1)
c.add_item(o2)
c.add_item(o3)
c.add_item(o4)
c.show()    
```

```python
# 9. Show aggregation with a Team class containing Player objects that can exist independently.
class Player:
    def __init__(self,jersy_no,player_name,quality):
        self.jersy_no=jersy_no
        self.player_name=player_name
        self.quality=quality

    def __repr__(self):
        return f"Jersy Number:{self.jersy_no} , Player name:{self.player_name} , Player Quality:{self.quality}"

class Team:
    def __init__(self,team_name):
        self.team_name=team_name
        self.player_list=[]

    def add_player(self,obj,limit=11):
        if len(self.player_list)<limit:
            if obj not in self.player_list:
                self.player_list.append(obj)
                #print(f"Player ={obj} added in the Team={self.team_name}")
            else:
                print(f"Player={obj} already in the Team={self.team_name}")
        else:
            raise ValueError(f"Team={self.team_name} is full now")
    def remove_player(self,obj):
        if obj in self.player_list:
            self.player_list.remove(obj)
            print(f"Player={obj} has removed from the Team={self.team_name}")
        else:
            print(f"No such player is in the squad")

    def show(self):
        if self.player_list:
            print("=======Player info======")
            for player in self.player_list:
                print(player)
        else:
            print("No player in the list")
p1=Player(45,"Rohit Sharma","Right Handed Batsman")
p2=Player(37,"Abhishek Sharma","Left Handed Batsman")
p3=Player(77,"Mohammad Siraj","Right Arm Fast Bowler")
t=Team("Indian Mens Cricket Team")
t.add_player(p1)
t.add_player(p2)
t.add_player(p3)
t.show()
```

```python
# 10. Create a Menu class that composes MenuItem objects recursively
class MenuItem:
    def __init__(self,s_no,menu_item):
        self.s_no=s_no
        self.menu_item=menu_item
    def __repr__(self):
        return f"{self.s_no}.{self.menu_item}"

class Menu:
    def __init__(self,menu_name):
        self.menu_name=menu_name
        self.menu_list=[]
    def add_menu(self,obj):
        if obj not in self.menu_list:
            self.menu_list.append(obj)
        else:
            return f"{obj} already exist in MenuItem"
            
    def remove_menu(self,obj):
        if obj in self.menu_list:
            self.menu_list.remove(obj)
        else:
            raise ValueError(f"{obj} not found in MenuItem")
            
    def show(self, indent=0):
        print(" " * indent + f" {self.menu_name}")
        for menu in self.menu_list:
            if isinstance(menu, Menu):
                menu.show(indent + 4)   # Recursive call for submenus
            else:
                print(" " * (indent + 4) + str(menu))
m1=MenuItem(1,"View Food Court")
m2=MenuItem(2,"View Price")
m3=MenuItem(3,"Place Order")
m4=MenuItem(4,"Add More Order")
m5=MenuItem(5,"Cancel Order")

starter_menu = Menu("Starters")
starter_menu.add_menu(MenuItem(1, "Paneer Tikka"))
starter_menu.add_menu(MenuItem(2, "Spring Rolls"))
starter_menu.add_menu(MenuItem(3, "Garlic Bread"))

main_course_menu = Menu("Main Course")
main_course_menu.add_menu(MenuItem(1, "Butter Chicken"))
main_course_menu.add_menu(MenuItem(2, "Veg Biryani"))
main_course_menu.add_menu(MenuItem(3, "Paneer Butter Masala"))

dessert_menu = Menu("Desserts")
dessert_menu.add_menu(MenuItem(1, "Gulab Jamun"))
dessert_menu.add_menu(MenuItem(2, "Rasmalai"))
dessert_menu.add_menu(MenuItem(3, "Ice Cream"))


list_menu_obj=[m1,m2,m3,m4,m5]
main_menu=Menu("View Options For Food ")
for m in list_menu_obj:
    main_menu.add_menu(m)

main_menu.add_menu(starter_menu)
main_menu.add_menu(main_course_menu)
main_menu.add_menu(dessert_menu)
print("All MenuItem object added to the Menu")
main_menu.show()
```
